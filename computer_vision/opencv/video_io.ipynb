```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load the video file
cap = cv2.VideoCapture('people.mp4')

# Check if the video opened successfully
if not cap.isOpened():
    print("Error: Could not open video.")
    exit()

# Desired output resolution
target_width = 960
target_height = 540

while cap.isOpened():
    ret, frame = cap.read()

    if not ret:
        print("No more frames or error reading the video.")
        break

    # Resize the frame to 1080p
    frame_resized = cv2.resize(frame, (target_width, target_height))

    # Show the resized video frame
    cv2.imshow('1080p Video', frame_resized)

    # Press 'q' to quit playback
    if cv2.waitKey(1) == ord('q'):
        break

# Release resources
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2

def filter(image):
    # Convert to grayscale
    img_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    
    # Apply Gaussian Blur
    img_blur = cv2.GaussianBlur(img_gray, (5, 5), 0)
    
    # Detect edges
    C_edges = cv2.Canny(img_blur, 10, 70)
    
    # Invert thresholded edges to create a sketch effect
    _, mask = cv2.threshold(C_edges, 70, 255, cv2.THRESH_BINARY_INV)
    
    return mask

# Open the video file
cap = cv2.VideoCapture('people.mp4')

# Target resolution
target_width = 960
target_height = 540

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Resize the frame
    frame_resized = cv2.resize(frame, (target_width, target_height))

    # Apply filter and display
    cv2.imshow('Filtered Video (960x540)', filter(frame_resized))

    # Exit on Enter key (ASCII 13)
    if cv2.waitKey(1) == 13:
        break

cap.release()
cv2.destroyAllWindows()
```
