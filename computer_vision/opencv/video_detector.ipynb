```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Open the video
cap = cv2.VideoCapture('people.mp4')

# Get video width and height
w = int(cap.get(3))  # CAP_PROP_FRAME_WIDTH
h = int(cap.get(4))  # CAP_PROP_FRAME_HEIGHT

# Output video writer
Out = cv2.VideoWriter('detections.mp4', cv2.VideoWriter_fourcc(*'mp4v'), 30, (w, h), True)

# Load Haar Cascade for face detection
face_detector = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

# Function to detect faces with adjusted parameters
def detect_faces(image):
    face_detections = face_detector.detectMultiScale(
        image,
        scaleFactor=1.2,      # Slightly more conservative scaling
        minNeighbors=8,       # Suppress false positives
        minSize=(99, 99)      # Ignore small detections
    )
    return face_detections

# Process each frame
while cap.isOpened():
    ret, frame = cap.read()
    if ret:
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        detection = detect_faces(gray)
        for (x, y, w, h) in detection:
            cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)
        Out.write(frame)
    else:
        break

# Release resources
cap.release()
Out.release()
cv2.destroyAllWindows()
```
